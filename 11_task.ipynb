```python
def calculate_discount_percentage(price):

    if price > 5000:
        return 30
    else:
        return 0
product_price = 1800000
discount = calculate_discount_percentage(product_price)
print(f"Product Price: {product_price:}")
print(f"Discount: {discount}%")
```

```python
def square(num):
  return num*num
print(square(5))
```

```python
def get_even(my_list):
    result = []
    for i in my_list:
        if i % 2 == 0:
            result.append(i)
    return result
print(get_even([1,2,3,4,5,6]))
```

```python
def reverse_string(text):
    return text[::-1]
print(reverse_string('hello'))
```

```python
def is_positive(num):
  return num > 0
print(is_positive(5))
print(is_positive(-5))
```

```python
def average(my_list):
  return sum(my_list)/len(my_list)
print(average([1,2,3,4,5,6]))
```

```python
def show_details(**kwargs):
  for key,value in kwargs.items():
    print(f"{key}: {value}")
show_details(Name="charukh",Branch="ECE",City="Chennai")
```

```python
def count_vowels(text):
  vowels = "aeiouAEIOU"
  count = 0
  for char in text:
    if char in vowels:
      count += 1
  return count
print(count_vowels("Generative AI"))
```

```python
def final_price(price,discount_percent):
  discount = price * discount_percent / 100
  return price - discount
print(final_price(1000,20))
```
