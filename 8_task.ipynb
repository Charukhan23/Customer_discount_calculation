```python
correct_pin = "1234"
attempts = 0

while attempts < 3:
    pin = input("PIN : ")
    if pin == correct_pin:
        print("Access Granted!")
        break
    else:
        attempts += 1
        print(f"Wrong! {3-attempts} chance left")
else:
    print("Card Blocked!")
```

```python
transactions = [100, -50, 200, -10, 500, 0]
valid = []
invalid = []
for t in transactions:
    if t > 0: # amount >0 na than valid
        valid.append(t)
    else:
        invalid.append(t)
print(f"Valid: {valid}")
print(f"Invalid: {invalid}")
```

```python
secret = 50
guesses = 0
while True:
    guess = int(input("Enter your guess: "))
    guesses = guesses + 1
    if guess > secret:
        print("Too High!")
        print()
    elif guess < secret:
        print("Too Low!")
        print()
    else:
        print("Congratulations! You guessed it!")
        print()
        break
print(f"You guessed the secret number in {guesses} moves.")
```
